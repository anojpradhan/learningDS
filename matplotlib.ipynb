Univariate Analysis: 
Here we only analyze one single data or feature

```python
data={ "Salary": [25000, 26000, 27000, 28000, 28000, 30000, 31000, 35000, 37000, 39000, 45000, 48000, 52000, 52000, 55000]}
```

```python
df= pd.DataFrame(data)
df.head()
```

```python
df.shape
```

```python
# Line Plot
plt.plot(df["Salary"], color="red", marker="*", linestyle=":", linewidth="1")
plt.grid()
plt.show()
```

```python
plt.hist(df['Salary'], bins=5, color="green")
plt.show()
```

```python
# df.loc[15]=0
plt.boxplot(df["Salary"])
plt.show()
```

Univariate Categorical columns

```python
df["dept"]=["IT","HR","Sales","Finance","IT","Sales","HR","HR","Sales","Finance","Sales","Finance","HR","IT","Sales"]
df.head()
```

```python
# Pie Chart
count= df['dept'].value_counts()
plt.pie(count, labels=count.index, autopct="%1.2f",explode=[0.1,0,0,0])
plt.axis("equal")
plt.show()
```

```python
# countplot

plt.bar(count.index, count, color=['green','black','red','blue'])
plt.show()
```

```python
# Bivariate Analysis- numeric columns

df["Age"]=[20, 23, 27, 28, 29, 32, 32, 33, 34, 34, 45, 45, 45, 56, 67]
# df.shape
df.head()
```

```python
# scatter plot
plt.scatter(df["Age"], df['Salary'], color='orange')
plt.show()
```

```python
# line plot 
plt.plot(df["Age"],df["Salary"], color='red', marker='o', linestyle="--")
plt.grid()
plt.show()
```

```python
# Bar chart 
plt.bar(df['Age'], df['Salary'], color='green', align="center")
plt.show()
```

Bivariate Analysis (Numerical - Categorical)

```python
hr_salaries= df[df["dept"]== "HR"]["Salary"]
hr_salaries
```

```python
it_salaries= df[df["dept"]== "IT"]["Salary"]
it_salaries
```

```python
sales_salaries= df[df["dept"]== "Sales"]["Salary"]
sales_salaries
```

```python
fin_salaries= df[df["dept"]== "Finance"]["Salary"]
fin_salaries
```

```python
# boxplot 
plt.boxplot([hr_salaries,fin_salaries,it_salaries,sales_salaries], label=['Hr','finance','IT','Sales'], orientation='horizontal')
plt.legend()
plt.show()
```

```python
# piechart
salary_by_dept= df.groupby("dept")["Salary"].sum()
salary_by_dept
```

```python
plt.pie(salary_by_dept,labels= salary_by_dept.index, autopct="%1.2f")
plt.axis("equal")
plt.show()
```

```python
# bar plot
hr_sal_mean= sum(hr_salaries)/len(hr_salaries)
it_sal_mean= sum(it_salaries)/len(it_salaries)
sales_sal_mean= sum(sales_salaries)/len(sales_salaries)
fin_sal_mean= sum(fin_salaries)/len(fin_salaries)
plt.bar(["HR","IT","Sales","Finance"], [hr_sal_mean,it_sal_mean,sales_sal_mean,fin_sal_mean], color=['green','red','black','blue'])
plt.grid()
plt.show()
```

Multivariate analysis: numeric columns

```python
df.head()
```

```python
df['Exp']= [1,1,1,2,2,2,2,3,3,4,4,4,5,5,5]
```

```python
df.head()
```

```python
df.shape
```

```python
# Bubble Plot
plt.scatter(df["Age"], df["Salary"],s= df["Exp"]*30,edgecolors="black" ,color="red")
plt.title("Age vs Salary vs Experience")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
# 2 numeric amd 1 categorical
colors= {"HR":"yellow","IT":"blue","Finance":"red","Sales":"black"}
for dept,color in colors.items():
    df_dept=df[df['dept']==dept]
    plt.scatter(df_dept["Age"],df_dept["Salary"],c=color, label=dept)
plt.legend()
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```
