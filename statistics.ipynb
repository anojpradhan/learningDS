Mean , Median , Mode and Standard Deviation

Mean: This is the average of set of numbers

Median: It is the middle value when data is arranged in the order

Mode: It is the most repeated value in the data set

Standard Deviation: This tells us how spread out the numbers are in a dataset. If it is low, that means data points are close to mean and if it is high they are spread

Variance: It also measures the spreadness of data, and is similar to sd, but variance is average squared distance and sd is square root of the variance.So it is harder to interpret in real worls terms, as SD brings that number back to readable everyday scale

## Distributions

It shows hoe data values are spread across different outcomes

There are two types of distributions.
1. Normal Distrubutions
2. Binomial Distributions

Normal Distribution is a bell shaped curve where most values lie near the average, and fewer values appear towards the extremes, commonly seen in heights, exam scores, etc.

Binomial Distrubution is used when there are only two possible outcomes like success/ failure, heads/tails, and it calculates the probability the certain number of successes over a fixed number of trails

## Histogram
It shows how often values fall within specific ranges.
For example, student marks can be grouped into ranges like 60-70 or 70-80 to easily see patterns and data spread.
