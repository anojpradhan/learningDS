```python
data={
    'id':[1,2,3,4,5,4],
    'name':["hari","ram","sita","alex","merry","elsa"],
    'age':[24,25,23,21,None,23],
    'rank':[3,2,5,4,1,6],
    'score':[80,90,65,79,97,63]
}
df= pd.DataFrame(data)
print(df)
```

```python
df.to_csv('students.csv', index= False)
students= pd.read_csv('students.csv')
print(students.head(2))
```

```python
students[['id','rank']]
```

```python
high_scorers= students[students['score']>80]
sorted_high_scorers= high_scorers.sort_values(by='rank', ascending=True)
print(sorted_high_scorers)
```
