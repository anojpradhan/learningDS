Linear Regression: 
It is a supervised ML algorithm used to predict continuous numerical values by learning the relationship between input variable and output variables.
It models the relationship between independent variable(input) and a dependent variable(target)
Assumes a linear relationship between inputs and outputs
Uses training data to learn the best fit line

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
data= {'Size': [1000,1500,2000,2500,3000], 'price':[200000,250000,300000,350000,400000]}
df= pd.DataFrame(data)
```

```python
model = LinearRegression()
model.fit(df[['Size']], df[['price']])
predicted_price= model.predict([[2200]])
predicted_price
```

```python
plt.scatter(df['Size'], df['price'],color='blue', label='Data points')
plt.plot(df['Size'], model.predict(df[['Size']]), color='red', label='Regression Line')
plt.title("Linear regression(Hourse Size vs Price) ")
plt.xlabel('Size of the house (sq ft)')
plt.ylabel('Price ($)')
plt.show()
```
