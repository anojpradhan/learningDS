Learning Seaborn here

```python
import seaborn as sns 
import pandas as pd 
import numpy as np 
```

```python
sns.get_dataset_names()
```

```python
penguins= sns.load_dataset("penguins")
penguins.head()
```

```python
penguins["species"].value_counts()
```

```python
penguins["island"].value_counts()
```

```python
penguins["sex"].value_counts()
```

```python
# Scatter plot
sns.set_style("whitegrid")
sns.set_context("talk")
sns.scatterplot(data=penguins, x="flipper_length_mm", y="body_mass_g", hue="island")

# this shows higher the flipper_length, higher the body mass and also dream and torgerson has lesser body mass and flipper length
```

```python
# palette
sns.scatterplot(data=penguins, x="flipper_length_mm", y="body_mass_g", hue="sex", palette='Dark2')
```

```python
sns.scatterplot(data=penguins, x='species', y='body_mass_g', hue='island', style='sex', alpha=0.5)
```

```python
sns.stripplot(data=penguins, x="species", y="body_mass_g", hue="island",jitter=True, dodge=True)
```

```python
# Swarm plot 
sns.swarmplot(data= penguins, x="species", y="body_mass_g", hue="island")
```

```python
# historgam
sns.set_context("notebook")
sns.set_style("white")
sns.histplot(data=penguins, x='body_mass_g', hue="sex", multiple="stack")
```

```python
# regression plot
sns.regplot(data=penguins, x="body_mass_g", y="flipper_length_mm")
```

```python
# line plot
sns.lineplot(data=penguins, x="body_mass_g", y="flipper_length_mm")
```
