Data visualization basically mean to show data using graphs and charts

We have two libraries in python , they are matpolib and seaborn. 
Matplotlib is very flexible, creates almost any type of chart, and is best for custom and detailed plots, but needs more code for styling.
Seaborn is built on the top of matpolib and it makes beautiful and complex charts with lesser amount of code


## Bar Chart
It helps to compare quantities or frequencies across independent categories

```python
import matplotlib.pyplot as plt
students=["Ram","Hari","Shyam","Kishan","Madhav"]
marks=[80,79,56,75,94]
```

```python
plt.bar(students, marks)
plt.xlabel('Students')
plt.ylabel('Marks')
plt.title("marks of students")
```

# Line graph
It helps to learn the change or trends

```python
import matplotlib.pyplot as plt
days=['Monday','Tuesday','Wednesday','Thursday','Friday']
traffic=[120,150,180,220,190]
plt.plot(days, traffic)
plt.title("Website Traffic over the week")
plt.xlabel("days")
plt.ylabel('Traffic')
```
