```python
import pandas as pd 
df = pd.DataFrame({
    'Age':[20,50,11],
    'Salary':[200000,30000,15000],
    'City':["ktm","pkh","brt"]
})
print(df)
```

### Here age , and salary are very differen values meaning that they are really far from each other , so what we would do is feature scaling

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df[['Age',"Salary"]]= scaler.fit_transform(df[['Age', 'Salary']])
print(df)
```
