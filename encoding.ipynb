ML models cannot understand text , we need to feed them number so using encoding we shall convert text like data to number format. Basically we need to convert categorical data into number. Some of methods of encoding are :
1. One hot encoding 
2. label Encoding 

```python
import pandas as pd 
df = pd.DataFrame({
    'Age':[20,50,11],
    'Salary':[200000,30000,15000],
    'City':["ktm","pkh","brt"]
})
print(df)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df[['Age',"Salary"]]= scaler.fit_transform(df[['Age', 'Salary']])
print(df)
```

Example of one hot encoding

```python
df= pd.get_dummies(df, columns=['City'], dtype=int)
print(df)
```

Example of label encoding

```python
from sklearn.preprocessing import LabelEncoder
df2 = pd.DataFrame({
    'Age':[20,50,11],
    'Salary':[200000,30000,15000],
    'City':["ktm","pkh","brt"]
})
print(df2)
```

```python
label_encoder = LabelEncoder()
df2['City']= label_encoder.fit_transform(df2['City'])
print(df2)
```
