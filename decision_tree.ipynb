## Decision Tree 
A decision tree is a classification model that predicts a category or a class. It works like a flowchart where it asks yes/ no questions.

```python
from sklearn.tree import DecisionTreeClassifier
import pandas as pd
data = { 'age':[25,30,35,40,45], 'income':[30000,40000,50000,60000, 700000], 'bought':[0,1,1,0,1]}
df= pd.DataFrame(data)
```

```python
model = DecisionTreeClassifier()
X= df[['age','income']]
y= df['bought']
model.fit(X,y)
predicted= model.predict([[36, 50000]])
print('Prediction(0= No , 1= Yes): ', predicted)
```
